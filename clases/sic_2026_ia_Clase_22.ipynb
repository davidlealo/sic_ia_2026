## 3. Por qué presentar con estructura de pitch

La atención de quien escucha es **intermitente**: a los cinco o diez minutos de una exposición sin estructura se empieza a perder gente, y quien no entiende se pierde cada vez más.

La estructura de pitch evita ese círculo vicioso porque cada parte prepara la siguiente y mantiene el interés:

1. **Gancho** que capte la atención.
2. **Problema**, con datos que muestren por qué importa.
3. **Solución** (y, si aporta, otras soluciones que existen).
4. **Prototipo** funcionando.
5. **Cierre**.

Recuerden que la presentación final dura **máximo 10 minutos** y que deben poder explicar el código (ver Clase 20).
